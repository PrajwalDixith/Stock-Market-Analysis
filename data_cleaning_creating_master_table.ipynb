```python
import os
import pandas as pd

# Directory where your stock files are located
base_dir = r"C:\aFiles And Media\DataAnalytics\HexaSolutions\Hexa Assignments\Assignments Till 09-10-2026\Assignment FINAL - PROJECTS -20260510T122027Z-3-001\Assignment FINAL - PROJECTS\7- SQL Stock Marketing Assignment\SQL - Stock Market Analysis Project"

# Mapping raw CSV filenames to stock names
stock_files = {
    'Bajaj Auto.csv': 'Bajaj Auto',
    'Eicher Motors.csv': 'Eicher Motors',
    'Hero Motocorp.csv': 'Hero Motocorp',
    'Infosys.csv': 'Infosys',
    'TCS.csv': 'TCS',
    'TVS Motors.csv': 'TVS Motors'
}

clean_frames = []

for file_name, stock_name in stock_files.items():
    file_path = os.path.join(base_dir, file_name)
    
    if not os.path.exists(file_path):
        print(f"Skipping {file_name}: file not found at {file_path}")
        continue
        
    df = pd.read_csv(file_path)
    
    # 1. Convert Date to standard SQL format (YYYY-MM-DD)
    df['Date'] = pd.to_datetime(df['Date'], format='%d-%B-%Y').dt.strftime('%Y-%m-%d')
    
    # 2. Add Stock_Name column
    df['Stock_Name'] = stock_name
    
    # 3. Rename Close Price to Closing_Price
    df = df.rename(columns={'Close Price': 'Closing_Price'})
    
    # 4. Filter only required columns specified in Part 2
    df_clean = df[['Stock_Name', 'Date', 'Closing_Price']].copy()
    
    # 5. Sort chronologically from earliest to latest
    df_clean = df_clean.sort_values('Date').reset_index(drop=True)
    
    # Correct file saving inside the target directory
    clean_path = os.path.join(base_dir, f"clean_{file_name}")
    df_clean.to_csv(clean_path, index=False)
    print(f"Saved: {clean_path} ({len(df_clean)} rows)")
    
    clean_frames.append(df_clean)

# Merge all into one consolidated master CSV
master_df = pd.concat(clean_frames, ignore_index=True)
master_path = os.path.join(base_dir, "master_stock_prices.csv")
master_df.to_csv(master_path, index=False)

print("\n--- Summary ---")
print(f"Master file saved: {master_path}")
print(f"Total rows: {len(master_df)} (889 rows * 6 stocks)")
print(f"Total null values: {master_df.isnull().sum().sum()}")
```
